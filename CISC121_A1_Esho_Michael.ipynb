````python
import java.util.Arrays;

class Solution {

    public String longestCommonPrefix(String[] strs) {
         String shorterWord;
        String longerWord;

        // Step 1: Sort the array alphabetically
        Arrays.sort(strs);

        // Get first and last words
        String firstWord = strs[0];
        String lastWord = strs[strs.length - 1]

        // Step 2: Find which word is shorter

        if (firstWord.length() <= lastWord.length()) {
            shorterWord = firstWord;
            longerWord = lastWord;
        } else {
            shorterWord = lastWord;
            longerWord = firstWord;
        }

        // Step 3: Convert both words into character arrays
        char[] shorterArray = shorterWord.toCharArray();
        char[] longerArray = longerWord.toCharArray();

        // Step 4: Go through both arrays saving common letters
        StringBuilder commonLetters = new StringBuilder();
        // string builder so i can add on to it

        // Loop until we reach the last letter of the shortest array
        for (int i = 0; i < shorterArray.length; i++) {
            char letterFromShorter = shorterArray[i];
            char letterFromLonger = longerArray[i];

            // Check if letters match
            if (letterFromShorter == letterFromLonger) {
                commonLetters.append(letterFromShorter);
            } else {
                //no match stopping the comparison
                break;
            }
        }

        // Step 5: Return the letters they have in common
        return  commonLetters.toString();
    }
}

````

````python
class Solution:
    def longestCommonPrefix(self, strs):
        # Step 1: Sort the array alphabetically
        strs.sort()

        # Get first and last words
        firstWord = strs[0]
        lastWord = strs[-1]

        # Step 2: Find the minimum length between the first and last words
        min_len = min(len(firstWord), len(lastWord))

        # Step 3: Find the common prefix
        common_prefix = ""
        for i in range(min_len):
            if firstWord[i] == lastWord[i]:
                common_prefix += firstWord[i]
            else:
                break

        # Step 4: Return the common prefix
        return common_prefix
````

and although they used c++ I did understand the solution, they counted the number of zeros and and stored all the non zero elements in another array, I wasnt sure how they added back the zeros and non zeros back in because they used a Push_back function so instead i thought of swaping them along with counting and saving the position of the zeros and non zeros



````python
class Solution(object):
    def moveZeroes(self, nums):
        #  Find all zero positions
        zero_positions = []
        for i in range(len(nums)):
            if nums[i] == 0:
                zero_positions.append(i)

        #  Find positions at the end where zeros should go
        num_zeros = len(zero_positions)
        end_positions = []
        for i in range(len(nums) - num_zeros, len(nums)):
            end_positions.append(i)

        #  Swap each zero with its corresponding end position
        for i in range(num_zeros):
            zero_pos = zero_positions[i]
            end_pos = end_positions[i]
            nums[zero_pos] = nums[end_pos]
            nums[end_pos] = nums[zero_pos]
        return nums
````
